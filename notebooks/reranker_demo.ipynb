```python
import pandas as pd
import numpy as np
import sys
import os

# Set path for src
sys.path.append(os.path.abspath('../src'))

# Import reranker
from reranker import Reranker

# Load processed playlist data
df_playlists = pd.read_pickle('../data/processed_playlists.pkl')  # playlist-track mapping
df_tracks = pd.read_pickle('../data/processed_tracks.pkl')      # track metadata

# === Build track popularity (count appearances in playlists) ===
track_popularity = df_playlists['track_uri'].value_counts()

# === Build track_artist_map dict ===
track_artist_map = pd.Series(df_tracks.artist_name.values, index=df_tracks.track_uri).to_dict()

# === Initialize reranker ===
reranker = Reranker(track_popularity=track_popularity, popularity_weight=0.7, diversity_weight=0.3)
reranker.set_artist_map(track_artist_map)

# === Sample usage ===
# Suppose you have some recommended tracks and scores (e.g., from hybrid recommender)
sample_recommended_tracks = ['track_uri_1', 'track_uri_2', 'track_uri_3']  # Replace with real URIs
sample_scores = [0.9, 0.85, 0.7]  # Dummy scores, usually from hybrid recommender

# Run reranker
reranked_tracks = reranker.rerank(sample_recommended_tracks, sample_scores)
print("Original recommended tracks:", sample_recommended_tracks)
print("Reranked tracks:", reranked_tracks)
```
